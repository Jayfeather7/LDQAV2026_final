Object: speech verbs in 'Crime and Punishment' by F. M. Dostoevsky.

Hyp1: since C&M, as (any) other works by Dostoevsky, is "male-heavy", when it comes to its character ensemble, the speech verbs that accompany the utterances by female characters could stand out from those by male characters (in frequency? semantics?)

Hyp2: speech verbs play a crucial role in characterization, so the individual use of speech verbs in the given mode of speech defined by the narrative context (dialogue, speaking to oneself, train of thought) is character-particular (in general and dependent on the addressee).

Source: .xml file with a tagged text (mode of speech, speaker, addressee)
Algorithms: extract speech verbs exclusively from the author's remarks (inquits) preceding or following direct speech, classify them into classes matching up with a pre-fab lexicon (with subsequent manual clean-up and completion), scores the most frequent ones (overall for masc and femn, per speaker and per addressee), calculates the shares of verbs per modes of speech

Results: Hyp1 disproved (data are skewed and should be carefully normalized, plots 1-4), Hyp2 verified (see the plots 11-19)
